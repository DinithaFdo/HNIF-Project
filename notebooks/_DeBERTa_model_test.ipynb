```python
import os
import sys

sys.path.append(os.path.abspath(".."))

import torch
from transformers import AutoModelForSequenceClassification, AutoTokenizer

print("Environment setup complete.")
print(f"CUDA Available: {torch.cuda.is_available()}")
```

```python
# Load a pre-trained DeBERTa-v3 model
MODEL_NAME = "vraj33/ai-text-detector-deberta"

print(f"Loading model and tokenizer: {MODEL_NAME}...")
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSequenceClassification.from_pretrained(
    MODEL_NAME, output_attentions=True
)

device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)
model.eval()

print(f"Model successfully loaded on device: {device}")
```

```python
from hnif import run_hnif_analysis

# Sample forensic test sentence
test_text = (
    "The algorithmic progression of artificial intelligence systems demonstrates a "
    "remarkable capacity for generating syntactically uniform prose."
)

# Simulated detection inputs (representing Ishara's classifier output)
simulated_prediction = "AI-Generated"
simulated_confidence = 0.984

print("Executing HNIF Forensic Pipeline...")
results = run_hnif_analysis(
    model=model,
    tokenizer=tokenizer,
    text=test_text,
    predicted_class=simulated_prediction,
    confidence=simulated_confidence,
)

print("Pipeline execution completed successfully!")
```

```python
import json

print("JSON Response Payload (Next.js Format):")
print("-" * 60)
print(json.dumps(results, indent=2))
print("-" * 60)
```
